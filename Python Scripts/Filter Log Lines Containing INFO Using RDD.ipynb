```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
                    .master("local") \
                    .appName("Filter Log Lines Containing INFO Using RDD") \
                    .getOrCreate()
```

```python
rdd = spark.sparkContext.textFile("../Data/Logs.txt")
```

```python
info_rdd = rdd.filter(lambda log: "INFO" in log)
```

```python
print("Info logs:", info_rdd.collect())
```
