```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
                    .appName("Survey Answer Frequency Analysis via Spark RDDs") \
                    .master("local") \
                    .getOrCreate()
```

### A call centre wants to know the number of occurrences of words 'satisfied' or 'good' in their C-SAT survey. 

```python
survey_rdd = spark.sparkContext.textFile('../Data/Survey.csv')
```

```python
survey_rdd.collect()
```

```python
pairs = survey_rdd.map(lambda row: row.split(",")) \
                    .map(lambda fields: (fields[0], fields[1]))
pairs.collect()
```

```python
pairs.map(lambda row: (row[1], 1)).reduceByKey(lambda x, y: x + y).collect()
```
