```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
                    .appName("RDD from a text file") \
                    .master("local[*]") \
                    .getOrCreate()
```

```python
spark
```

```python
customer_rdd = spark.sparkContext.textFile(name="../Data/Customer Data.txt", 
                                           minPartitions=12)
```

```python
customer_rdd
```

```python
print("Number of partitions:", customer_rdd.getNumPartitions())
```

```python
print("First 10 rows:", customer_rdd.take(10))
```

```python
print("Count of rows:", customer_rdd.count())
```

```python
print("First row:", customer_rdd.first())
```
